```python
#Q35. 
# Prediction:[[1,99],[3,4]]
import copy 
a = [[1, 2], [3, 4]] 
b = copy.copy(a) 
b[0][1] = 99 
print(a) 
#Explanation: copy.copy() creates a shallow copy of the list a. This means that the outer list is copied, but the inner lists are still referencing the same objects.modifying b[0][1] chnages inner list of a as well.

#Verification:
```

```python
#Q36. 
# Prediction: [[1,2],[3,4]]
import copy 
a = [[1, 2], [3, 4]] 
b = copy.deepcopy(a) 
b[0][1] = 99 
print(a)
#Explanation: copy.deepcopy(a) creates a deep copy of the list a. This means that both the outer and inner lists are copied, so modifying b does not affect a.

#Verification:
```
