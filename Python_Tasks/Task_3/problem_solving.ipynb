Q1. Find the Frequency of Each Word in a Sentence
Write a Python program to count how many times each word appears in a given sentence and store the result in a dictionary.

Example Input:
python is easy and python is powerful

Expected Output:
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

**Pseudocode**

1. Start with an empty dictionary.
2. Split the sentence into individual words.
3. Visit each word one by one.
4. If the word already exists in the dictionary, increase its count by one.
5. Otherwise, add the word to the dictionary with a count of one.
6. Return and display the dictionary.

```python
def freq(sentence):
    frequency={}

    for i in sentence.split():
        if i in frequency:
            frequency[i]+=1
        else:
            frequency[i]=1
    return frequency


sentence1="python easy and python is powerful"
print(freq(sentence1))

# Additional test case
sentence1="sql is useful and sql is easy to learn"
print(freq(sentence1))
```

Q2. Find the Frequency of Each Character in a Word
Write a Python program to count how many times each character appears in a given word and store the result in a dictionary.

Example Input:
banana

Expected Output:
{'b': 1, 'a': 3, 'n': 2}

**Pseudocode**

1. Start with an empty dictionary.
2. Visit each character.
3. Increase its count if it already exists; otherwise, initialize it to 1.
4. Return the dictionary.

```python
def char_freq(word):
    frequency1={}
    for char in word:
        if char in frequency1:
            frequency1[char]+=1
        else:
            frequency1[char]=1
    return frequency1

word="banana"
print(char_freq(word))

# Additional test case
word="canada"
print(char_freq(word))
```

Q3. Find the First Non-Repeating Character in a Word
Write a Python program to find the first character that appears only once in a given word. If no such character exists, return None.

Example Input:
swiss

Expected Output:
w

**Pseudocode**
1. Count the frequency of every character.
2. Visit the characters in their original order.
3. Return the first character whose frequency is one.
4. If none exists, return None.

```python
def first_non_repeating(word):
    frequency=char_freq(word)
    for char in word:
        if frequency[char]==1:
            return char
    return None

word="swiss"
print(first_non_repeating(word))

# Additional test case
word="aadhya"
print(first_non_repeating(word))
```

Q4. Check Whether Two Strings Are Anagrams
Write a Python program to check whether two strings are anagrams.
Two strings are anagrams if they contain the same characters with the same frequencies, regardless of their order.

Example Input:
s1 = "listen"
s2 = "silent"

Expected Output:
True

**Pseudocode**
1. Convert both strings to lowercase and remove spaces.
2. Count the characters in both strings.
3. Compare the frequency dictionaries.
4. Return True if they match; otherwise, return False.


```python
def anagram(s1,s2):
    s1=s1.replace(" ","").lower()
    s2=s2.replace(" ","").lower()
    return char_freq(s1)==char_freq(s2)

print(anagram("listen","silent"))

# Additional test case
print(anagram("earth","heart"))
```

Q5. Find the First Repeating Character in a String
Write a Python program to find the first character whose second occurrence is encountered while reading the string from left to right.
If no character repeats, return None.

Example Input:
programming

Expected Output:
r

**Pseudocode**
1. Define the function for first repeating character
2. Remove spaces and change it to lower case.
3. Create an empty set
4. Check each character and if it is occuring once, add it to set , else return the character.
5. Finally return none if char is not repeated.

```python
def first_repeating(word1):
    word=word1.replace(" ","").lower()
    fir_char=set()
    for char in word:
        if char in fir_char:
            return char
        
        fir_char.add(char)
    return None

word1="programming"
print(first_repeating(word1))

# Additional test case
word1="sense"
print(first_repeating(word1))
```

Q6. Move All Zeros to the End of a List
Write a Python program to move all zeros to the end of a list while maintaining the original order of non-zero elements.

Example Input:
[0, 3, 0, 1, 2, 0, 4]

Expected Output:
[3, 1, 2, 4, 0, 0, 0]

**Pseudocode**
1. Create an empty list for non-zero value
2. check for each number whether it is non-zero. If it is non-zero then add it to the list.
3. Subtract the actual list with the new list. It will give the count of zeros.
4. Then add zeros to the new list with number of counts and print the result.

```python
def list_of_nums(nums):
    non_zeros=[]
    for num in nums:
        if num!=0:
            non_zeros.append(num)

    zero_num=len(nums)-len(non_zeros)

    return non_zeros+[0]*zero_num

print(list_of_nums([0, 3, 0, 1, 2, 0, 4]))
# Additional test case
print(list_of_nums([0, 0, 5, 7, 0, 9, 0]))
```

Q7. Find the Second Highest Distinct Number in a List
Write a Python program to find the second-largest distinct number in a given list.
Duplicate values should not be considered separate ranks.
If a second distinct number does not exist, return None.

Example Input:
[10, 20, 40, 30, 40, 20]

Expected Output:
30

**Pseudocode**
1. Create an empty set for storing unique numbers
2. check if the set has less than 2 numbers return none, else sort the set in ascending order and return the last 2nd element to get the second highest number
3. print the result.

```python
def second_largest_distinct(lst):
    unique_nums=set(lst)
    if(len(unique_nums)<2):
        return None
    return sorted(unique_nums)[-2]

lst=[10, 20, 40, 30, 40, 20]
print(second_largest_distinct(lst))
# Additional test case
lst=[300,700,400,100,100,500]
print(second_largest_distinct(lst))
```

Q8. Remove Duplicates Without Changing the Original Order
Write a Python program to remove duplicate elements from a list while preserving the order of their first occurrence.

Example Input:
[4, 2, 4, 5, 2, 7, 5]

Expected Output:
[4, 2, 5, 7]

**Pseudocode**
1. Create an empty set
2. create an empty list to store the result
3. check for each number, and add the number in the set. if it is not in the set add the number in the list.
4. print the result to get the list without duplicates.

```python
def remove_duplicates(lst):
    list1=set()
    result=[]

    for item in lst:
        if item not in list1:
            list1.add(item)
            result.append(item)
    return result
lst=[4,2,4,5,2,7,5]
print(remove_duplicates(lst))
# Additional test case
lst=[12,14,12,13,17,14]
print(remove_duplicates(lst))
```

Q9. Find the Missing Number from 1 to N
Write a Python program to find the missing number in a list containing numbers from 1 to N.
The list contains exactly one missing number, and all remaining numbers appear only once.

Example Input:
nums = [1, 2, 4, 5, 6]
N = 6

Expected Output:
3

**Pseudocode**
1. For finding sum of n consecutive numbers n*(n+1)/2 is used.
2. Actual sum is calculated using sum() function.
3. expected - actual sum is subtracted to find the missing number.
4. Print the missing number.

```python
def missing_nums(nums,n):
    expected_sum=n*(n+1)/2
    actual_sum=sum(nums)

    return int(expected_sum - actual_sum)

print(missing_nums([1,2,4,5,6],6))

# Additional test case
print(missing_nums([1,3,4],4))
```

Q10. Find Two Numbers Whose Sum Equals a Target
Write a Python program to find the first pair of numbers whose sum equals a given target.
Return the pair as a list. If no valid pair exists, return an empty list.
Return the pair whose second element is encountered earliest when scanning the input from left to right. If multiple earlier elements qualify, choose the earliest one.

Example Input:
nums = [2, 7, 11, 15]
target = 9

Expected Output:
[2, 7]

**Pseudocode**
1. Visit each number from left to right.
2. Calculate the number needed to reach the target.
3. If that number has appeared earlier, return the pair.
4. Otherwise, record the current number.
5. Return an empty list if no pair exists.

```python
def two_sum(nums, target):
    s=set()

    for num in nums:
        required_num=target-num

        if required_num in s:
            return [required_num,num]

        s.add(num)
    return []

nums=[2, 7, 11, 15]
target=9
print(two_sum(nums,target))

# Additional test case
nums=[3,10,5,9]
target=15
print(two_sum(nums,target))
```

Q11. Find the Number Closest to Zero
Write a Python program to find the number closest to zero in a given list.
If two numbers are equally close to zero, return the positive number.

Example Input:
[-8, -3, 2, 5, -2]

Expected Output:
2

**Pseudocode**
1. Compare numbers using their absolute values.
2. When distances are equal, prefer the positive number.
3. Return the selected number.

```python
def closest_to_zero(nums):
    if not nums:
        return None
    return min(nums,key=lambda num:(abs(num),-num))

nums=[-8, -3, 2, 5, -2]
print(closest_to_zero(nums))

# Additional test case
nums=[6, -1, 1, -9, 3]
print(closest_to_zero(nums))
```

Q12. Find Common Elements Between Two Lists
Write a Python program to find the common elements between two lists.
Each common element should appear only once in the result, maintaining the order of its first occurrence in the first list.

Example Input:
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

Expected Output:
[3, 4, 5]

**Pseudocode**
1. Record the elements of the second list in a set.
2. Visit the first list in order.
3. Add elements that exist in both lists, only once.
4. Return the result.

```python
def common_elements(list1,list2):
    s=set(list2)
    result=[]
    s1=set()

    for num in list1:
        if num in s and num not in s1:
            result.append(num)
            s1.add(num)
    return result

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
print(common_elements(list1,list2))

# Additional test case
list1 = [20, 30, 40, 50, 60]
list2 = [30, 40, 50, 60, 70]
print(common_elements(list1,list2))
```

Q13. Rotate a List to the Right by K Positions
Write a Python program to rotate the elements of a list to the right by K positions.
Elements shifted beyond the last position should move to the beginning of the list.

Example Input:
nums = [1, 2, 3, 4, 5]
K = 2

Expected Output:
[4, 5, 1, 2, 3]

**Pseudocode**
1. create an empty list.
2. Reduce K using the length of the list.
3. take the last K elements and place them first.
4. Append the remaining elements.

```python
def rotate(nums,K):
    if not nums:
        return []
    K=K%len(nums)
    return nums[-K:]+nums[:-K] if K else nums[:]

nums = [1, 2, 3, 4, 5]
K = 2
print(rotate(nums,K))
# Additional test case
nums = [13,16,7,14,20]
K = 3
print(rotate(nums,K))
```

Q14. Find the Maximum Sum of Two Consecutive Elements
Write a Python program to find the maximum sum of any two consecutive elements in a list.
Assume the list contains at least two elements.

Example Input:
[4, 7, 2, 9, 5]

Expected Output:
14

**Pseudocode**
1. Calculate the sum of the first two elements.
2. Visit each consecutive pair.
3. Update the maximum whenever a larger sum appears.
4. Return the maximum sum.

```python
def max_consecutive_sum(nums):
    max_sum=nums[0]+nums[1]

    for i in range(1,len(nums)-1):
        current_sum=nums[i]+nums[i+1]

        if current_sum>max_sum:
            max_sum=current_sum
    return max_sum

nums=[4, 7, 2, 9, 5]
print(max_consecutive_sum(nums))
# Additional test case
nums=[1, 2, 3, 4, 5]
print(max_consecutive_sum(nums))
    
```

Q15. Find All Pairs of Numbers Whose Sum Equals a Target
Write a Python program to find all unique pairs of numbers whose sum equals a given target.
Each pair should contain two distinct list positions, and duplicate pairs should not be included.
Return each pair in ascending numerical order. Return the list of pairs sorted by their first element. If no valid pairs exist, return an empty list.

Example Input:
nums = [2, 4, 3, 5, 7, 8, 1]
target = 9

Expected Output:
[[1, 8], [2, 7], [4, 5]]

**Pseudocode**
1. check every pair of distinct list positions.
2. check whether their sum equals the target.
3. Sort the two values in each valid pair.
4. Remove duplicate pairs and sort the final result.

```python
def all_pairs(nums,target):
    pairs=set()
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                pair=tuple(sorted((nums[i],nums[j])))
                pairs.add(pair)
    return [list(pair) for pair in sorted(pairs)]

nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
print(all_pairs(nums,target))
# Additional test case
nums = [1,1,2,2,3,4]
target = 5
print(all_pairs(nums,target))
```
